```python
import telegram
from telegram.ext import Updater, CommandHandler, MessageHandler, Filters

# Put the token that BotFather gave you
TOKEN = '7678392303:AAHp4tuM6gRkNOP3TbpOE2yvPwKZiGGY-B0'

# Start function
def start(update, context):
    update.message.reply_text("Ciao! Sono il tuo chatbot. Scrivimi qualcosa!")

# Message reply function
def handle_message(update, context):
    text = update.message.text.lower()

    if 'hi' in text:
        update.message.reply_text("Hi! How can I help you today?")
    elif 'prices' in text:
        update.message.reply_text("Our prices start at 45€.")
    else:
        update.message.reply_text("Sorry, I didn't understand. Can you repeat?")

# Setup updater and dispatcher
updater = Updater(TOKEN, use_context=True)
dp = updater.dispatcher

dp.add_handler(CommandHandler('start', start))
dp.add_handler(MessageHandler(Filters.text, handle_message))

# Starting the bot
updater.start_polling()
updater.idle()
```
